## 1. 專案簡介

PromptBench 是一個以臺灣繁體中文呈現的生成式 AI 工具與 Prompt 知識庫。網站不是單純的工具排行榜，而是把任務、工具限制、官方來源與可重複使用的 Prompt 整理成同一套工作流程。

第一版收錄 7 項 AI 工具與 6 份任務型 Prompt，提供搜尋、分類篩選、兩項工具比較、工具與 Prompt 詳細頁、複製 Prompt、選擇指南及資料方法說明。

## 5. 心得與反思

這次製作 PromptBench，讓我理解到使用生成式 AI 完成專案的重點，不只是請 AI 快速產生網頁，而是先把專案目的、資料格式、視覺風格與開發流程說清楚。建立 AGENTS.md 後，AI 在後續新增工具、Prompt 與互動功能時，可以遵循相同規則，成果也比較一致。

實際協作過程中，我也發現 AI 的輸出仍需要人工檢查。工具資訊可能改版，連結、引用與功能描述都必須回到官方來源確認；搜尋、篩選、比較及手機版畫面也需要實際操作，不能只相信程式已經完成。這讓我體會到 AI 比較適合作為協作夥伴，而不是完全取代人的判斷。

這次專案最大的收穫，是把一次性的網頁作業轉變成可以持續延伸的系統。未來只要依照既有資料結構新增內容，列表、詳細頁、搜尋、篩選與比較功能就能繼續使用。若再進一步改進，我會增加更多實際測試紀錄、工具更新歷史與 Prompt 使用成效，讓網站的內容更可靠、更有參考價值。

## 6. 簡短說明：規則如何幫助 AI 延伸專案

AGENTS.md 定義專案目的、繁體中文用語、視覺系統、程式規範，以及工具與 Prompt 的固定資料格式。新的 AI Agent 只要先閱讀該文件，就能沿用相同欄位、分類與共用元件新增內容，不需要為每項資料另外建立頁面。文件亦規定來源查證、版本紀錄與交付檢查流程，使後續修改能保持一致並接受檢查。
